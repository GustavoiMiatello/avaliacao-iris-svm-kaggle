## **1. Carregando os dados e preparando**

```python
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split

iris = load_iris()
X = iris.data
y = iris.target

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.20, random_state=42)
```

```python
from sklearn.model_selection import GridSearchCV
from sklearn.metrics import accuracy_score, classification_report
from sklearn.tree import DecisionTreeClassifier

# Árvore de Decisão sem GridSearchCV
dtc = DecisionTreeClassifier(random_state=42)

# treinar o modelo
dtc.fit(X_train, y_train)

# predição do modelo
y_pred_dtc = dtc.predict(X_test)

print(f'Acurácia inicial Árvore de Decisão: {accuracy_score(y_pred_dtc, y_test) * 100}%')

# 1. Definir o espaço de busca dos hiperparâmetros
param_grid = {
    'criterion': ['gini', 'entropy', 'log_loss'],
    'max_depth': [None, 2, 3, 4, 5, 6],
    'min_samples_split': [2, 5, 10],
    'min_samples_leaf': [1, 2, 4]
}

# 2. Configurar o GridSearchCV
grid_search = GridSearchCV(
    estimator=dtc,
    param_grid=param_grid,
    cv=5,
    scoring='accuracy',
    n_jobs=-1
)

# 3. Treinar a busca nos dados de treino
grid_search.fit(X_train, y_train)

# 4. Visualizar os melhores hiperparâmetros encontrados
print("Melhores hiperparâmetros:", grid_search.best_params_)

# 5. Avaliar o melhor modelo no conjunto de teste
melhor_dtc = grid_search.best_estimator_
y_pred_grid = melhor_dtc.predict(X_test)

print("=== Random Forest ===")
print(f"Acurácia no teste: {accuracy_score(y_test, y_pred_grid) * 100}%\n")
print(classification_report(y_test, y_pred_grid, target_names=iris.target_names))
```

```python
from sklearn.ensemble import RandomForestClassifier

# Floresta Aleatória sem GridSearchCV
rf = RandomForestClassifier(random_state=42)

# treinar o modelo
rf.fit(X_train, y_train)

# predição do modelo
y_pred_rf = rf.predict(X_test)

# 1. Grade de hiperparâmetros
param_grid_rf = {
    'n_estimators': [50, 100, 200],
    'max_depth': [None, 3, 5, 8],
    'min_samples_split': [2, 5],
    'min_samples_leaf': [1, 2],
    'max_features': ['sqrt', 'log2']
}

# 2. Estimador base e GridSearch
rf = RandomForestClassifier(random_state=42)
grid_rf = GridSearchCV(
    estimator=rf,
    param_grid=param_grid_rf,
    cv=5,
    scoring='accuracy',
    n_jobs=-1
)

# 3. Treinamento
grid_rf.fit(X_train, y_train)

# 4. Avaliação
print("=== Random Forest ===")
print("Melhores parâmetros:", grid_rf.best_params_)

melhor_rf = grid_rf.best_estimator_
y_pred_rf = melhor_rf.predict(X_test)

print(f"Acurácia no teste: {accuracy_score(y_test, y_pred_rf) * 100}%")
print(classification_report(y_test, y_pred_rf, target_names=iris.target_names))
```

```python
from sklearn.ensemble import GradientBoostingClassifier

# Gradient Boosting sem GridSearchCV
gb = GradientBoostingClassifier(random_state=42)

# treinar o modelo
gb.fit(X_train, y_train)

# predição do modelo
y_pred_gb = gb.predict(X_test)

# 1. Grade de hiperparâmetros
param_grid_gb = {
    'n_estimators': [50, 100, 150],
    'learning_rate': [0.01, 0.05, 0.1, 0.2],
    'max_depth': [2, 3, 4],  # Árvores de boosting costumam ser mais rasas
    'subsample': [0.8, 1.0]   # Fração de amostras por árvore (stochastic gradient boosting)
}

# 2. Estimador base e GridSearch
gb = GradientBoostingClassifier(random_state=42)
grid_gb = GridSearchCV(
    estimator=gb,
    param_grid=param_grid_gb,
    cv=5,
    scoring='accuracy',
    n_jobs=-1
)

# 3. Treinamento
grid_gb.fit(X_train, y_train)

# 4. Avaliação
print("=== Gradient Boosting ===")
print("Melhores parâmetros:", grid_gb.best_params_)

melhor_gb = grid_gb.best_estimator_
y_pred_gb = melhor_gb.predict(X_test)

print(f"Acurácia no teste: {accuracy_score(y_test, y_pred_gb) * 100}%")
print(classification_report(y_test, y_pred_gb, target_names=iris.target_names))
```
